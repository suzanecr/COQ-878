```python
import os

import numpy as np
import matplotlib.pyplot as plt

from ase import Atoms, Atom
from ase.build import molecule, surface, bulk, add_adsorbate
from ase.io import write, read
from ase.visualize.plot import plot_atoms
from ase.constraints import FixAtoms
from ase.spacegroup import crystal
from ase.build import make_supercell
```

```python
layers = 2              # número de camadas da superfície
vacuum = 30             # Espaço de vácuo
distance = 3.0          # Distância inicial etanol-superfície
n_etoh = 3              # Número de moléculas de etanol
```

```python
hap = read("HAP.cif")

# SUPERFÍCIE (001) -- + comum
slab = surface(hap, (0, 0, 1), layers=layers)
```

```python
slab.pbc = True

min_z = np.min(slab.positions[:, 2])
slab.positions[:, 2] -= min_z

z_top_slab = np.max(slab.positions[:, 2])
new_c = z_top_slab + vacuum

cell = slab.get_cell()
cell[2, 2] = new_c
slab.set_cell(cell, scale_atoms=False)
```

# Inserção de UMA molécula de etanol. Se quiser mais de uma molécula rodar o código no próximo markdown sem rodar esse.

```python
ethanol = molecule('CH3CH2OH')
ethanol.translate(-ethanol.get_center_of_mass())
```

```python
ethanol.translate([slab.cell[0,0]/2,
                slab.cell[1,1]/2,
                z_top_slab + distance])
```

```python
system_total = slab + ethanol
system_total.pbc = True
```

```python
os.makedirs("sistema_HAP_EtOH", exist_ok=True)
write("sistema_HAP_EtOH/HAP_EtOH.pdb", system_total)
write("sistema_HAP_EtOH/HAP_EtOH.xyz", system_total)
write("sistema_HAP_EtOH/HAP_EtOH.cif", system_total)

print("\nArquivo gerado em sistema_HAP_EtOH/HAP_EtOH.pdb\n")
```

# Para inserção de mais de uma molécula de etanol ALEATORIAMENTE. Deve ser rodado logo após a criação da HAP.

```python
mol_list = []
for i in range(n_etoh):
    etoh = molecule('CH3CH2OH')
    etoh.translate(-etoh.get_center_of_mass())

    # Posição aleatória em XY e fixa em Z
    x_rand = np.random.rand() * cell[0,0]
    y_rand = np.random.rand() * cell[1,1]
    z_rand = z_top_slab + distance + np.random.rand() * (vacuum - distance)

    etoh.translate([x_rand, y_rand, z_rand])
    mol_list.append(etoh)


system_total = slab
for mol in mol_list:
    system_total += mol

system_total.pbc = True

os.makedirs("sistema_HAP_randEtOH", exist_ok=True)
write("sistema_HAP_randEtOH/HAP_randEtOH.pdb", system_total)
write("sistema_HAP_randEtOH/HAP_randEtOH.xyz", system_total)
write("sistema_HAP_randEtOH/HAP_randEtOH.cif", system_total)

print(f"Sistema gerado com {n_etoh} moléculas de etanol aleatórias!")
```
